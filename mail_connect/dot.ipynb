```python
from google.oauth2.credentials import Credentials
from google_auth_oauthlib.flow import InstalledAppFlow
from google.auth.transport.requests import Request
from googleapiclient.discovery import build
import pandas as pd
import os

# Define the scopes required by the Gmail API
SCOPES = ['https://www.googleapis.com/auth/gmail.readonly']

def authenticate_gmail():
    """Authenticate with Gmail API and return the service object."""
    creds = None
    token_path = 'token.json'  # Token file for storing authentication tokens
    credentials_path = 'credentials.json'  # Your credentials file from Step 1

    # Check if the token file already exists
    if os.path.exists(token_path):
        creds = Credentials.from_authorized_user_file(token_path, SCOPES)

    # If credentials are invalid or missing, start the OAuth flow
    if not creds or not creds.valid:
        if creds and creds.expired and creds.refresh_token:
            creds.refresh(Request())
        else:
            flow = InstalledAppFlow.from_client_secrets_file(credentials_path, SCOPES)
            creds = flow.run_local_server(port=0)
        
        # Save the credentials for future use
        with open(token_path, 'w') as token_file:
            token_file.write(creds.to_json())

    # Build the Gmail service object
    return build('gmail', 'v1', credentials=creds)

def fetch_emails_by_label(service, label_name):
    """Fetch emails with a specific label and return as a list of dictionaries."""
    # Get the list of all labels
    labels = service.users().labels().list(userId='me').execute().get('labels', [])
    label_id = None

    # Find the label ID for the specified label name
    for label in labels:
        if label['name'].lower() == label_name.lower():
            label_id = label['id']
            break

    if not label_id:
        print(f"Label '{label_name}' not found!")
        return []

    # Fetch emails with the specific label
    messages = service.users().messages().list(userId='me', labelIds=[label_id]).execute().get('messages', [])
    email_data = []

    for message in messages:
        message_id = message['id']
        # Fetch message details
        msg_details = service.users().messages().get(userId='me', id=message_id).execute()
        snippet = msg_details.get('snippet', '')
        headers = msg_details['payload'].get('headers', [])
        subject = next((header['value'] for header in headers if header['name'] == 'Subject'), 'No Subject')
        sender = next((header['value'] for header in headers if header['name'] == 'From'), 'Unknown Sender')
        
        email_data.append({'Message ID': message_id, 'Subject': subject, 'Sender': sender, 'Snippet': snippet})

    return email_data

def main():
    # Authenticate and build the Gmail service
    service = authenticate_gmail()

    # Specify the Gmail label you want to filter by
    label_name = "zet reports"  # Replace with your desired Gmail label

    # Fetch emails with the specified label
    email_data = fetch_emails_by_label(service, label_name)

    if email_data:
        # Convert to Pandas DataFrame
        df = pd.DataFrame(email_data)

        # Display the DataFrame
        print(df)

        # Save the DataFrame to a CSV file (optional)
        df.to_csv('emails_by_label.csv', index=False)
        print("Data saved to 'emails_by_label.csv'")
    else:
        print(f"No emails found with label '{label_name}'")

if __name__ == '__main__':
    main()
```

```python
from google.oauth2.credentials import Credentials
from google_auth_oauthlib.flow import InstalledAppFlow
from google.auth.transport.requests import Request
from googleapiclient.discovery import build
import pandas as pd
import os
import base64
import sqlite3
import io  # Add this import for BytesIO

# Define the scopes required by the Gmail API
SCOPES = ['https://www.googleapis.com/auth/gmail.readonly']

def authenticate_gmail():
    """Authenticate with Gmail API and return the service object."""
    creds = None
    token_path = 'token.json'  # Token file for storing authentication tokens
    credentials_path = 'credentials.json'  # Your credentials file from Step 1

    # Check if the token file already exists
    if os.path.exists(token_path):
        creds = Credentials.from_authorized_user_file(token_path, SCOPES)

    # If credentials are invalid or missing, start the OAuth flow
    if not creds or not creds.valid:
        if creds and creds.expired and creds.refresh_token:
            creds.refresh(Request())
        else:
            flow = InstalledAppFlow.from_client_secrets_file(credentials_path, SCOPES)
            creds = flow.run_local_server(port=0)
        
        # Save the credentials for future use
        with open(token_path, 'w') as token_file:
            token_file.write(creds.to_json())

    # Build the Gmail service object
    return build('gmail', 'v1', credentials=creds)

def fetch_emails_with_label(service, label_name):
    """Fetch all emails with a specific label and return message details."""
    # Get the list of all labels
    labels = service.users().labels().list(userId='me').execute().get('labels', [])
    label_id = None

    # Find the label ID for the specified label name
    for label in labels:
        if label['name'].lower() == label_name.lower():
            label_id = label['id']
            break

    if not label_id:
        print(f"Label '{label_name}' not found!")
        return []

    # Fetch emails with the specific label
    messages = []
    response = service.users().messages().list(userId='me', labelIds=[label_id]).execute()
    messages.extend(response.get('messages', []))
    while 'nextPageToken' in response:
        response = service.users().messages().list(userId='me', labelIds=[label_id], pageToken=response['nextPageToken']).execute()
        messages.extend(response.get('messages', []))
    
    return messages

def extract_csv_attachments(service, message):
    """Download CSV attachments from a Gmail message."""
    attachments = []
    message_details = service.users().messages().get(userId='me', id=message['id']).execute()
    parts = message_details['payload'].get('parts', [])

    for part in parts:
        if part['filename'].endswith('.csv'):
            attachment_id = part['body']['attachmentId']
            attachment = service.users().messages().attachments().get(
                userId='me', messageId=message['id'], id=attachment_id).execute()
            file_data = base64.urlsafe_b64decode(attachment['data'])
            attachments.append((part['filename'], file_data))
    
    return attachments



def save_csv_to_db(csv_data, db_name="emails.db"):
    """Save CSV data into a SQLite database."""
    conn = sqlite3.connect(db_name)

    for filename, data in csv_data:
        # Convert bytes to a file-like object
        csv_file = io.BytesIO(data)
        
        # Create a Pandas DataFrame
        df = pd.read_csv(csv_file)
        table_name = os.path.splitext(filename)[0]  # Use the filename as the table name
        
        # Save DataFrame to SQLite
        df.to_sql(table_name, conn, if_exists='replace', index=False)
        print(f"Saved {filename} to database as table '{table_name}'.")

    conn.close()


def main():
    # Authenticate and build the Gmail service
    service = authenticate_gmail()

    # Specify the Gmail label you want to filter by
    label_name = "zet reports"  # Replace with your desired Gmail label

    # Fetch emails with the specified label
    messages = fetch_emails_with_label(service, label_name)

    if not messages:
        print(f"No emails found with label '{label_name}'")
        return

    # Extract CSV attachments from emails
    all_csv_data = []
    for message in messages:
        csv_attachments = extract_csv_attachments(service, message)
        if csv_attachments:
            all_csv_data.extend(csv_attachments)

    if all_csv_data:
        # Save CSV data to a SQLite database
        save_csv_to_db(all_csv_data)
        print("All CSV files have been processed and stored in the database.")
    else:
        print(f"No CSV attachments found in emails with label '{label_name}'.")

if __name__ == '__main__':
    main()
```
