```python
# Creating a dictionary
student = {
    "name": "Aditi",
    "age": 22,
    "major": "Computer Science",
    "is_graduated": False
}

# Accessing values
print(student["name"])      # Output: Aditi
print(student["age"])       # Output: 22

# Adding a new key-value pair
student["university"] = "IIT Delhi"

# Updating a value
student["age"] = 23

# Removing a key
del student["is_graduated"]

# Printing the whole dictionary
print(student)
```

```python
def countdown(n):
    while n > 0:
        print(n)
        n -= 1
    print("Time's up!")

# Call the function
countdown(3)
```
