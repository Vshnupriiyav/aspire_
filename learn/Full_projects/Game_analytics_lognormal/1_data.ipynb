```python
# get cohort month (y-m format), get activity_month (y-m fomart), get_period (cohort_month-activity month)
```

```python
Pl_activity_ca.dtypes
```

```python
Pl_activity_ca["ActivityMonth"] = pd.to_datetime(Pl_activity_ca["ActivityMonth"])
Pl_activity_ca["Signup_Date"] = pd.to_datetime(Pl_activity_ca["Signup_Date"])
```

```python
Pl_activity_ca["Cohort_month"] =  Pl_activity_ca["Signup_Date"].dt.strftime("%y-%m")
```

```python
## fucntion to help us get month, year separatly, which will be used to calculate period

def get_m_y(df,column):
    year = df[column].dt.year
    month = df[column].dt.month
    return year, month
```

```python
act_year, act_month = get_m_y(Pl_activity_ca,"ActivityMonth")
coh_year, coh_month = get_m_y(Pl_activity_ca,"Signup_Date")
```

```python
Pl_activity_ca["period"] = (act_year-coh_year)*12 + (act_month-coh_month)
```

```python
Pl_activity_ca.head(2)
```

```python
coh_by_month1 = Pl_activity_ca.groupby(["Cohort_month", "period"]).agg(
    user_count=("src_player_id", "count"),
    total_revenue=("revenue", "sum")
).reset_index()
```

```python
initial_users = initial_users.rename(columns={"user_count":"initial_users"})
```

```python
coh_by_month1 = coh_by_month1.merge(initial_users, on="Cohort_month")
```

```python
coh_by_month1.head(1)
```

```python
coh_by_month1["percent_users"] = (coh_by_month1["user_count"]/coh_by_month1["initial_users"])*100
```

```python
# x is the column (or series) that will be provided, it will return, year, month and day as default 1

def get_month(x):
    return dt.datetime(x.year, x.month, 1)
```
