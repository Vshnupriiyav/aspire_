```python
pivots = df.pivot_table("Temperature", index="City", columns="Year")
pivots
```

```python
pivots.index
```

```python
#Get the mean temp by year

mean_temp_by_year = pivots.mean()
mean_temp_by_year
```

```python
mean_temp_by_year.max()
```

```python
# Filter for the year that had the highest mean temp
print(mean_temp_by_year[mean_temp_by_year == mean_temp_by_year.max()])
```

```python
# Get the mean temp by city
mean_by_city = pivots.mean(axis="columns")
mean_by_city
```

```python
# Filter for the city that had the lowest mean temp

print(mean_by_city[mean_by_city == mean_by_city.min()])
```
